```python
# Import libraries
import numpy as np
import matplotlib.pyplot as plt
from numpy import random

# Generate a seed value for a random number of generator
random.seed(0)

# total indicates the total no of people is in each age range
total={20:0, 30:0, 40:0, 50:0, 60:0}
#purchases indicates the total no of purchases were made by each person in these age range
purchases={20:0, 30:0, 40:0, 50:0, 60:0}
# total number of purchases regardless of age
totalPurchases=0

# create a loop to iterate through 50000 radom people.
for _ in range(50000):
  ageDecade=random.choice([20, 30, 40, 50, 60])
  # Compute a purchase probability given by the age
  purchaseProbability=float(ageDecade) / 100.0
  # collect the total for the generated age decade
  total[ageDecade] +=1
  # check the person bought something if random number is less than the purchase probability
  if(random.random() < purchaseProbability):
    # total number of purchases that are done across the entire dataset
    totalPurchases +=1
    # track of purchases for the individual age decade
    purchases[ageDecade] +=1
```

Conditional Probability

```python
# compute P(A|B), where A is "purchase" and B is "the person in it's 50's"
PAB=float(purchases[50])/float(total[50])
print('P(purchase|50s): ' +str(PAB))

# compute P(B) means the probability of being 50 in the dataset
PB= float(total[50])/50000.0
print('P(50s): ' +str(PB))

# compute p(A) means the overall probability of buying something, regardless of its age
PA= float(totalPurchases)/50000.0
print('P(perchase): ' +str(PA))
```

```python
#  P(A|B) = P(A,B)/ P(A)

# Compute P(A,B) where it is the probability of both being in its 50's and buying something P(A,B) out of the total population
print('P(50s, purchase ): ' +str(float(purchases[50]) / 50000.0))
```

```python
# compute the overall probability of buying something and the overall probability of being in 50's
print('P(50s) P(purchase): ' +str(PA*PB))
```

```python
# Compute P(A|B)
print((purchases[50]/50000.0) / PB)
```
